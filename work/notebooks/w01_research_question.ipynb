## 2. The question: decision, action, cost of a wrong call

*What decision does your work improve? Who acts on it? What does a wrong recommendation cost?*

## Research Question

### Decision
Which content pages should be prioritized for review and refresh?

### Action
The SEO or content team reviews the highest-ranked pages and decides whether to update, expand, improve, or monitor them.

### Cost of a Wrong Recommendation
If the model recommends the wrong pages, the team may waste time improving pages that were already performing well while missing pages that genuinely needed attention. This reduces efficiency and may result in lost search traffic.

### Why Machine Learning?
Large websites can contain thousands of pages. Reviewing every page manually is expensive and time-consuming. Machine learning can rank pages according to their likelihood of needing attention, helping teams focus their efforts where they are most valuable.
